# Аппроксимация распределения двухточечным распределением

Поставим задачу конструирования 3х точечного распределения с заданными $\mu, D, S, K$

$\mu=u_1p_1+u_2p_2+vq$

Матожидание остается таким же из-за конструкции

$\Large
p_i=\frac{1+\mu}{1+u_i}w_i
$

где 

$\Large
w_i=\frac{u_i^{-T}}{\sum\limits_ju_j^{-T}}
$

$\Large
w_i < 1
$

$\Large
w_1 + w_2 = 1
$

$v=-1$

$q=1-p_1-p_2$

остальные моменты обозначим их как центральные но они не явлются центральными

$\Large
\begin{aligned}
D &= u_1^2 p_1 + u_2^2 p_2 + v^2 q, \\
S &= u_1^3 p_1 + u_2^3 p_2 + v^3 q, \\
K &= u_1^4 p_1 + u_2^4 p_2 + v^4 q.
\end{aligned}
$

Упрощение для $D$:

$\Large
\begin{aligned}
D &= u_1^2 p_1 + u_2^2 p_2 + 1 - p_1 - p_2, \\
\frac{D - 1}{1+\mu} &= (u_1-1) w_1 + (u_2-1) w_2 \\
\frac{D - 1}{1+\mu} + 1 &= u_1w_1 + u_2w_2 \\
D' &= \frac{D - 1}{1+\mu} + 1
\end{aligned}
$

Выразим из уравнения явно T

$\Large
\begin{aligned}
D'(u_1^{-T}+u_2^{-T}) &= u_1u_1^{-T} + u_2u_2^{-T} \\
D'(u_1^{-T}+u_2^{-T}) &= u_1u_1^{-T} + u_2u_2^{-T} |: u_1^{-T} \\
D'\left(1+\left(\frac{u_1}{u_2}\right)^{T}\right) &= u_1 + u_2\left(\frac{u_1}{u_2}\right)^{T} \\
D'\left(1+\left(\frac{u_1}{u_2}\right)^{T}\right) &= u_1 + u_2\left(\frac{u_1}{u_2}\right)^{T} \\
D'+D'\left(\frac{u_1}{u_2}\right)^{T} &= u_1 + u_2\left(\frac{u_1}{u_2}\right)^{T} \\
\left(D'-u_2\right)\left(\frac{u_1}{u_2}\right)^{T} &= u_1 - D' \\
\left(\frac{u_1}{u_2}\right)^{T} &= \frac{u_1 - D'}{D'-u_2} \\
T\ln\left(\frac{u_1}{u_2}\right) &= \ln\frac{u_1 - D'}{D'-u_2} \\
T &= \frac{\ln\frac{u_1 - D'}{D'-u_2}}{\ln\left(\frac{u_1}{u_2}\right)} \\
\end{aligned}
$


Упрощение для $S$:

$\Large
\begin{aligned}
S &= u_1^3 p_1 + u_2^3 p_2 + v^3 q, \\
S &= u_1^3 p_1 + u_2^3 p_2 - 1 + p_1 + p_2, \\
S &= (u_1^3+1) p_1 + (u_2^3 + 1) p_2 - 1, \\
S &= (u_1^2 - u_1 + 1) w_1 + (u_2^2 - u_2 + 1) w_2 - 1, \\
\frac{S-1}{1+\mu} - 1 &= (u_1^2 - u_1) w_1 + (u_2^2 - u_2) w_2, \\
\frac{S-1}{1+\mu} - 1 + \frac{D - 1}{1+\mu} + 1 &= u_1^2 w_1 + u_2^2 w_2, \\
\frac{S-1}{1+\mu} + \frac{D - 1}{1+\mu} &= u_1^2 w_1 + u_2^2 w_2, \\
\frac{D+S-2}{1+\mu} &= u_1^2 w_1 + u_2^2 w_2, \\
S' &= \frac{D+S-2}{1+\mu}, \\
\end{aligned}
$

Выражаем $T$:

$\Large
\begin{aligned}
S' &= u_1^2 w_1 + u_2^2 w_2, \\
S'(u_1^{-T}+u_2^{-T}) &= u_1^2 u_1^{-T} + u_2^2 u_2^{-T}, \\
S'\left(1+\left(\frac{u_1}{u_2}\right)^{T}\right) &= u_1^2  + u_2^2 \left(\frac{u_1}{u_2}\right)^{T}, \\
S'+S'\left(\frac{u_1}{u_2}\right)^{T} &= u_1^2  + u_2^2 \left(\frac{u_1}{u_2}\right)^{T}, \\
(S'-u_2^2)\left(\frac{u_1}{u_2}\right)^{T} &= u_1^2  - S', \\
\left(\frac{u_1}{u_2}\right)^{T} &= \frac{u_1^2  - S'}{S'-u_2^2}, \\
T &= \frac{\ln\frac{u_1^2  - S'}{S'-u_2^2}}{\ln{\frac{u_1}{u_2}}}, \\
\end{aligned}
$

Упрощение для $K$:

$\Large
\begin{aligned}
K &= u_1^4 p_1 + u_2^4 p_2 + v^4 q, \\
K &= u_1^4 p_1 + u_2^4 p_2 + 1 - p_1 - p_2, \\
K &= (u_1^4 - 1) p_1 + (u_2^4 - 1) p_2 + 1, \\
\frac{K - 1}{1+\mu} &= (u_1^2 + 1)(u_1 - 1) w_1 + (u_2^2 + 1)(u_2 - 1) w_2, \\
\frac{K - 1}{1+\mu} &= (u_1^3 - u_1^2 + u_1 - 1) w_1 + (u_2^3 - u_2^2 + u_2 - 1) w_2, \\
\frac{K - 1}{1+\mu} &= u_1^3w_1 + u_2^3w_2 - S'+ D' - 1, \\
\frac{K - 1}{1+\mu} &= u_1^3w_1 + u_2^3w_2 - S'+ D' - 1, \\
D' &= \frac{D - 1}{1+\mu} + 1, \\
S' &= \frac{D+S-2}{1+\mu}, \\
\frac{K - 1}{1+\mu} &= u_1^3w_1 + u_2^3w_2, \\
\frac{K + S - 2}{1+\mu} &= u_1^3w_1 + u_2^3w_2, \\
K' &= \frac{K + S - 2}{1+\mu} \\
K' &= u_1^3w_1 + u_2^3w_2, \\
T &= \frac{\ln\frac{u_1^3  - K'}{K'-u_3^2}}{\ln{\frac{u_1}{u_2}}}, \\
\end{aligned}
$


Система

$\Large
\begin{aligned}
D' &= \frac{D - 1}{1+\mu} + 1, \\
S' &= \frac{D+S-2}{1+\mu}, \\
K' &= \frac{K + S - 2}{1+\mu} \\
\end{aligned}
$

$\Large
\begin{cases}
T = \dfrac{\ln\!\left(\dfrac{u_1 - D'}{D' - u_2}\right)}
         {\ln\!\left(\dfrac{u_1}{u_2}\right)}, \\[10pt]
T = \dfrac{\ln\!\left(\dfrac{u_1^2 - S'}{S' - u_2^2}\right)}
         {\ln\!\left(\dfrac{u_1}{u_2}\right)}, \\[10pt]
T = \dfrac{\ln\!\left(\dfrac{u_1^3 - K'}{K' - u_2^3}\right)}
         {\ln\!\left(\dfrac{u_1}{u_2}\right)}.
\end{cases}
$

Получим систему для нахождения $u_1, u_2$

$\Large
\begin{cases}
\dfrac{u_1 - D'}{D' - u_2} = \dfrac{u_1^2 - S'}{S' - u_2^2}, \\[10pt]
\dfrac{u_1 - D'}{D' - u_2} = \dfrac{u_1^3 - K'}{K' - u_2^3}.
\end{cases}
$

$
u_1>\max(D',\sqrt{S'},\sqrt[3]{K'})
$

$
u_2<\min(D',\sqrt{S'},\sqrt[3]{K'})
$

Для простоты переобозначим

$\Large
\begin{cases}

## Определение параметра $T$ для слота с заданными $\mu$ и $D=\sigma^2$


$u_i$ - коэффициенты выйгрышей

$v=-1$ - проигрыш

$\Large
p_i=\frac{1+\mu}{1+u_i}w_i
$ - вероятность выйгрышей

$\Large
w_i=\frac{u_i^{-T}}{\sum\limits_ju_j^{-T}}
$

$q=1-\sum\limits_i p_i$

Пусть известна дисперсия $D$


$D=\sum\limits_i u_i^2p_i+v^2q$

$D=u_1^2p_1+u_2^2p_2+u_n^2p_n+...+v^2q$

$D=u_1^2p_1+u_2^2p_2+...+u_n^2p_n+1-p_1-p_2-...-p_n$

$D=(u_1^2-1)p_1+(u_2^2-1)p_2+...+(u_n^2-1)p_n+1$

$D-1=(u_1^2-1)p_1+(u_2^2-1)p_2+...+(u_n^2-1)p_n$

$\frac{D-1}{1+\mu}=\frac{u_1^2-1}{u_1+1}w_1+\frac{u_2^2-1}{u_2+1}w_2+...+\frac{u_n^2-1}{u_n+1}w_n$

$\frac{D-1}{1+\mu}=(u_1-1)w_1+(u_2-1)w_2+...+(u_n-1)w_n$

$\frac{D-1}{1+\mu}=u_1w_1+u_2w_2+...+u_nw_n - 1$

$\frac{D-1}{1+\mu}+1=u_1w_1+u_2w_2+...+u_nw_n$

$\frac{D+\mu}{1+\mu}=u_1w_1+u_2w_2+...+u_nw_n$

$D'=\frac{D+\mu}{1+\mu}$

$D'(u_1^{-T}+u_2^{-T}+...+u_n^{-T})=u_1^{-T}+u_2^{-T}+...+u_n^{-T}$

$D'(u_1^{-T}+u_2^{-T}+...+u_n^{-T})=u_1u_1^{-T}+u_2u_2^{-T}+...+u_nu_n^{-T}$

$(D'-u_1)u_1^{-T}+(D'-u_2)u_2^{-T}+...+(D'-u_n)u_n^{-T}=0$

Таким образом задача свилась к нахождению корня $f(T^*)=0$ для функции:

$f(T)=(D'-u_1)u_1^{-T}+(D'-u_2)u_2^{-T}+...+(D'-u_n)u_n^{-T}$

в качестве начального приближения можем брать $T=1$





```python
import numpy as np
from scipy.optimize import root_scalar

def slot_wins(MaxWin=1000, use_loss_coef=False):
    x_win1 = np.arange(1, 10, 1)
    x_win2 = np.arange(10, 50, 1)
    x_win3 = np.arange(50, 100, 5)
    x_win4 = np.arange(100, 500, 10)
    x_win5 = np.arange(500, 1000, 100)
    x_win6 = np.arange(1000, 10000, 1000)
    x_win7 = np.arange(10000, 100_000, 10000)
    u_wins = np.concatenate((x_win1, x_win2, x_win3, x_win4, x_win5, x_win6, x_win7))
    if use_loss_coef:
        u_wins = np.concatenate([[-1],u_wins])
    u_wins = u_wins[u_wins <= MaxWin]
    return u_wins


def solve_T_scipy(u, D_, T0=1.0):
    u = np.asarray(u, float)


    def f(T):
        return np.sum((D_ - u) * u**(-T))

    # Метод секущих с начальной точкой
    sol = root_scalar(f, x0=T0, x1=T0+1, method='secant')

    return sol.root



D=10
mu=-0.05

D_=(D+mu)/(1+mu)

u = slot_wins(MaxWin=1000, use_loss_coef=False)

print("u:", u)

T = solve_T_scipy(u, D_, T0=1)

print("Solved T:", T)

w = u**(-T)/np.sum(u**(-T))
p = (1+mu)/(1+u)*w
q = 1 - np.sum(p)

print("D_check:", np.sum(u**2*p) + 1*q)
```
